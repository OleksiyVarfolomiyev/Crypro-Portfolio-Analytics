# IB account analytics

The IB counterpart of `hl_account.ipynb`: balances, today's PnL, open positions with ATR stop/target levels, current-book risk, and saved fills for one IB account. Read-only: the TWS connection cannot place orders.

Needs TWS (or IB Gateway) running and logged in with the API enabled (**Global Configuration → API → Settings → Enable ActiveX and Socket Clients**, port 7496 live / 7497 paper). The account comes from `accounts.json`; set `ACCOUNT` to choose another.

```python
from importlib import reload
import desk.ib.account_analytics as implementation
reload(implementation)  # pick up code changes without restarting the kernel

ACCOUNT = None  # an IB account number or accounts.json label; None uses the configured IB account
PORT = None     # override the TWS socket port, e.g. 7497 for paper

analytics = implementation.IBAccountAnalytics(ACCOUNT, port=PORT)  # IB records are saved in ../trading-desk-data/ib
_ = analytics.fetch_data()
```

## Account, PnL and risk

IB's API has no account-value history, so beta, alpha, Sharpe, volatility, VaR and drawdown price **today's holdings** over the past year of IB daily closes against SPY. Ranges are 90% bootstrap ranges. Open risk is the same ATR stop measure the executors cap desk-wide.
